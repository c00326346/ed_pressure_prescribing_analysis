```python
import pandas as pd
```

```python
atc = pd.read_csv("../data/raw/Distribution of Medicines by ATC.csv")
benzo = pd.read_csv("../data/raw/Benzodiazepines Z Drugs Claims.csv")
trolley = pd.read_csv("../data/raw/Inmo Trolley Data.csv")
```

```python
print(atc.shape)
print(benzo.shape)
print(trolley.shape)
```

```python
display(atc.info())
display(benzo.info())
display(trolley.info())
```

```python
display(atc.isna().sum())
display(benzo.isna().sum())
display(trolley.isna().sum())
```

```python
atc[atc["Prescribing Frequency"].isna()].head(20)
```

```python
atc_clean = atc[atc["Prescribing Frequency"].notna()].copy()
atc_clean.shape
```

```python
atc_clean["YEAR_MONTH"] = pd.to_datetime(
    atc_clean["YEAR_MONTH"].astype(str),
    format="%Y%m"
)

atc_clean["YEAR_MONTH"].head()
```

```python
benzo["YEAR_MONTH"] = pd.to_datetime(
    benzo["YEAR_MONTH"].astype(str),
    format="%Y%m"
)

benzo["YEAR_MONTH"].head()
```

```python
trolley["YEAR_MONTH"] = pd.to_datetime(
    trolley["YEAR_MONTH"].astype(str),
    format="%Y%m"
)

trolley["YEAR_MONTH"].head()
```

```python
print("ATC:", atc_clean["YEAR_MONTH"].min(), "to", atc_clean["YEAR_MONTH"].max())
print("Benzo:", benzo["YEAR_MONTH"].min(), "to", benzo["YEAR_MONTH"].max())
print("Trolley:", trolley["YEAR_MONTH"].min(), "to", trolley["YEAR_MONTH"].max())
```

```python
atc_clean[["ATC Class Category", "ATC Class Code", "Product Category", "Prescribing Frequency"]].head(30)
```

```python
atc_monthly_wide = atc_clean.pivot_table(
    index="YEAR_MONTH",
    columns="ATC Class Code",
    values="Prescribing Frequency",
    aggfunc="sum"
).reset_index()

atc_monthly_wide.head()
```

```python
atc_monthly_wide.columns.name = None
atc_monthly_wide.head()
```

```python
master_atc_trolley = trolley.merge(
    atc_monthly_wide,
    on="YEAR_MONTH",
    how="inner"
)
master_atc_trolley.shape
```

```python
master_atc_trolley.head()
```

```python
master_atc_trolley.tail()
```

```python
master_atc_trolley.isna().sum().sort_values(ascending=False).head(20)
```

```python
atc_cols_to_drop = ["L03", "J07", "A08", "R02", "A16", "D04"]

master_atc_trolley_clean = master_atc_trolley.drop(columns=atc_cols_to_drop)

master_atc_trolley_clean = master_atc_trolley_clean.fillna(0)

master_atc_trolley_clean.shape
```

```python
master_atc_trolley_clean.isna().sum().sum()
```

```python
master_atc_trolley_clean.to_csv(
    "../data/processed/master_atc_trolley_clean.csv",
    index=False
)
```

```python
benzo.head()
```

```python
benzo_monthly_wide = benzo.pivot_table(
    index="YEAR_MONTH",
    columns="ATC Class",
    values="Prescribing Frequency",
    aggfunc="sum"
).reset_index()

benzo_monthly_wide.columns.name = None

benzo_monthly_wide.head()
```

```python
benzo_freq_wide = benzo.pivot_table(
    index="YEAR_MONTH",
    columns="ATC Class",
    values="Prescribing Frequency",
    aggfunc="sum"
).add_prefix("BENZO_FREQ_").reset_index()

benzo_patients_wide = benzo.pivot_table(
    index="YEAR_MONTH",
    columns="ATC Class",
    values="Number of Patients",
    aggfunc="sum"
).add_prefix("BENZO_PATIENTS_").reset_index()
```

```python
benzo_monthly_wide = (
    benzo_freq_wide
    .merge(benzo_patients_wide, on="YEAR_MONTH", how="inner")
)

benzo_monthly_wide.head()
```

```python
benzo_missing = benzo_monthly_wide.isna().sum().sort_values(ascending=False)
benzo_missing[benzo_missing > 0]
```

```python
benzo_cols_to_drop = benzo_missing[benzo_missing > 20].index.tolist()
benzo_monthly_clean = benzo_monthly_wide.drop(columns=benzo_cols_to_drop)
benzo_monthly_clean = benzo_monthly_clean.fillna(0)
```

```python
benzo_monthly_clean.shape
```

```python
benzo_monthly_clean.isna().sum().sum()
```

```python
master_benzo_trolley_clean = trolley.merge(
    benzo_monthly_clean,
    on="YEAR_MONTH",
    how="inner"
)

master_benzo_trolley_clean.shape
```

```python
display(master_benzo_trolley_clean.head())
display(master_benzo_trolley_clean.tail())
```

```python
master_benzo_trolley_clean.to_csv(
    "../data/processed/master_benzo_trolley_clean.csv",
    index=False
)
```
